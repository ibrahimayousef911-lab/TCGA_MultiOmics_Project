```python
import pandas as pd
import numpy as np

# Load the saved RNA-seq data
rna = pd.read_pickle("01_Data/rna.pkl")

# Check the shape
print("RNA-seq Shape:", rna.shape)

# Check for missing values
print("Missing values:", rna.isnull().sum().sum())

# Check for duplicate genes
print("Duplicate genes:", rna.index.duplicated().sum())
```

```python
# Check if data is already log-transformed
print("Max value:", rna.max().max())
print("Min value:", rna.min().min())
```

```python
# Calculate variance for each gene
gene_variance = rna.var(axis=1)

# Select top 5000 most variable genes
top_genes = gene_variance.nlargest(5000).index

# Subset the RNA-seq data
rna_top = rna.loc[top_genes]

print("Top 5000 genes shape:", rna_top.shape)
```

```python
# Save the top genes
rna_top.to_pickle("01_Data/rna_top5000.pkl")

print("Saved rna_top5000.pkl!")
```

```python
# Load the methylation data
methyl = pd.read_pickle("01_Data/methyl.pkl")

# Calculate variance for each CpG site
cpg_variance = methyl.var(axis=1)

# Select top 10000 most variable CpG sites
top_cpgs = cpg_variance.nlargest(10000).index

# Subset the methylation data
methyl_top = methyl.loc[top_cpgs]

print("Top 10000 CpGs shape:", methyl_top.shape)
```

```python
# Save the top CpGs
methyl_top.to_pickle("01_Data/methyl_top10000.pkl")

print("Saved methyl_top10000.pkl!")
```

```python
# Load the mutation data
mutation = pd.read_pickle("01_Data/mutation.pkl")

# Check the columns
print(mutation.columns)

# Count mutations per gene (if the column is 'gene')
if 'gene' in mutation.columns:
    gene_mutation_counts = mutation['gene'].value_counts()
    top_mutated_genes = gene_mutation_counts.nlargest(100).index
    print("Top 100 mutated genes:", top_mutated_genes[:10])
else:
    print("No 'gene' column found. Check the columns.")
```

```python
# Save the top mutated genes
top_mutated_genes = gene_mutation_counts.nlargest(100).index

# Save to a file
pd.Series(top_mutated_genes).to_csv("01_Data/top_mutated_genes.csv", index=False)

print("Saved top_mutated_genes.csv!")
print("Number of genes:", len(top_mutated_genes))
```

```python
# Save all preprocessed data
rna_top.to_pickle("01_Data/rna_top5000.pkl")
methyl_top.to_pickle("01_Data/methyl_top10000.pkl")
pd.Series(top_mutated_genes).to_csv("01_Data/top_mutated_genes.csv", index=False)

print("All preprocessed data saved!")
```

```python
# Load the preprocessed data
rna_top = pd.read_pickle("01_Data/rna_top5000.pkl")
methyl_top = pd.read_pickle("01_Data/methyl_top10000.pkl")

# Transpose (MOFA2 needs samples x features)
rna_t = rna_top.T
methyl_t = methyl_top.T

# Check
print("RNA shape (samples x genes):", rna_t.shape)
print("Methylation shape (samples x CpGs):", methyl_t.shape)
```

```python
# Find common samples
common_samples = rna_t.index.intersection(methyl_t.index)
print("Number of common samples:", len(common_samples))

# Subset both datasets to common samples
rna_common = rna_t.loc[common_samples]
methyl_common = methyl_t.loc[common_samples]

# Check
print("RNA shape:", rna_common.shape)
print("Methylation shape:", methyl_common.shape)
```

```python
# Save the matched data
rna_common.to_pickle("01_Data/rna_matched.pkl")
methyl_common.to_pickle("01_Data/methyl_matched.pkl")

print("Saved matched data!")
```
